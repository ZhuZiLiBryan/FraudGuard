```python
import duckdb

con = duckdb.connect()

con.execute("""
    COPY (
        SELECT *
        FROM read_csv('data/train_transaction.csv', sample_size=-1) AS transactions 
        LEFT JOIN read_csv('data/train_identity.csv', sample_size=-1) AS identity
        USING (TransactionID) 
        ORDER BY TransactionDT
    )
    TO 'data/processed/fraud_data.parquet'
    (FORMAT PARQUET, COMPRESSION SNAPPY)
""")

con.close()
```
