```python
import duckdb
```

```python
con = duckdb.connect()

# Get total number transactions, distinct transactions, range of datetime, average fraud rate
print(con.execute("""
    SELECT 
    COUNT(*) AS num_transactions,
    COUNT(DISTINCT TransactionID) AS distinct_transactions,
    MAX(TransactionDT) - MIN(TransactionDT) as dt_range,
    AVG(isFraud) AS fraud_rate
    FROM 'data/processed/fraud_data.parquet'
""").df())

print(con.execute("""
    SELECT 
    COUNT(*) as null_count
    FROM 'data/processed/fraud_data.parquet'
    WHERE DeviceType IS NOT NULL
""").df())

con.close()
```
